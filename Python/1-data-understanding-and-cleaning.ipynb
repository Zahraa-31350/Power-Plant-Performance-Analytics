```python
#  Total number of duplicated values
df.duplicated().sum()
```

## Data Quality Findings

The dataset was assessed for missing values and duplicate records.

No missing values were identified across all features, indicating that the dataset is complete and does not require missing value treatment.

However, 41 duplicate rows were detected. Since duplicate records do not provide additional information and may affect statistical analysis and machine learning model performance, they will be removed during the data cleaning stage.

## Outlier Detection

```python
Q1 = df.quantile(0.25)
Q3 = df.quantile(0.75)

IQR = Q3 - Q1

outliers = ((df < (Q1 - 1.5 * IQR)) | 
            (df > (Q3 + 1.5 * IQR))).sum()

outliers
```

## Outlier Detection Findings

The dataset was evaluated for potential outliers using the Interquartile Range (IQR) method.

No outliers were detected in Ambient Temperature (AT), Exhaust Vacuum (V), or Electrical Power Output (PE).

A small number of potential outliers were identified in Ambient Pressure (AP) and Relative Humidity (RH). These observations represent a very small percentage of the dataset and may correspond to valid operating conditions rather than data errors. Therefore, they will be retained for further analysis.

# 7- Data Cleaning

```python
# Tuple representing the dimensionality with duplicated values
df.shape
```

```python
# Remove duplication
df = df.drop_duplicates()

# Tuple representing the dimensionality without duplicated values
df.shape
```

## Data Cleaning Summary

As part of the data cleaning process, duplicate records were removed from the dataset. A total of *41 duplicate rows* were identified and deleted, reducing the dataset size from *9,568* to *9,527* observations.

No missing values were found, and potential outliers were retained because they likely represent valid operational conditions rather than data quality issues. The cleaned dataset is now ready for exploratory data analysis.

```python
# Rename columns for better readability
df.rename(
    columns={
        "AT": "Ambient_Temperature",
        "V": "Exhaust_Vacuum",
        "AP": "Ambient_Pressure",
        "RH": "Relative_Humidity",
        "PE": "Power_Output"
    },
    inplace=True
)

# Verify updated column names
df.columns
```
